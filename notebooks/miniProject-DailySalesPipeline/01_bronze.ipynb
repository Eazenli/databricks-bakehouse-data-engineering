```python
from pyspark.sql import functions as F 
```

```python
source_df = spark.read.table(source_table)

bronze_df = (
    source_df
    .withColumn('ingest_timestamp', F.current_timestamp())
    .withColumn('ingest_batch_id', F.lit('week01_inital_load'))
    .withColumn('source_table', F.lit(source_table))
)

bronze_df.write.mode('overwrite').format('delta').saveAsTable(bronze_table)
```

```python
# bronze table controle 

source_table_rows = source_df.count()
bronze_table_rows = spark.read.table(bronze_table).count()

assert source_table_rows == bronze_table_rows, "Erreur : Data in source table and bronze table are not equal."
```
