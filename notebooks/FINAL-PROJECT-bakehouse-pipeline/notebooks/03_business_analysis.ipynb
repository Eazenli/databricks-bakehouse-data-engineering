```python
from pyspark.sql import functions as F
from src.config import config

gold_df = spark.table(config.gold_daily_franchise_sales)
```

```python
gold_df.show(5)
```

#### Top 3 franchises by daily revenue

```python
display(
    gold_df.filter(F.col("rank_sum_revenue") <= 3)
    .select("transaction_date","rank_sum_revenue","franchiseID","franchiseName","sum_revenue","amount_of_customers")
    .orderBy("transaction_date","rank_sum_revenue")
)
```

#### Total daily revenue

```python
daily_revenue_df = (
    gold_df.groupBy("transaction_date")
    .agg(F.sum("sum_revenue").alias("daily_revenue"))
    .orderBy("transaction_date")
)
daily_revenue_df.display()
```

#### Revenue by franchise over the processed period

```python
display(
    gold_df.groupBy("franchiseID","franchiseName")
    .agg(F.sum("sum_revenue").alias("total_revenue"))
    .orderBy(F.desc("total_revenue"))
)
```
